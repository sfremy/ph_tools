## Question 1

<p>Implement the function <u>find_largest(numbers)</u>, so it returns the largest number from the list <u>numbers</u>.</p>
<p>Notes</p>

- <u>numbers</u> contains only integers.
- <u>numbers</u> always contains at least one element, and never contains more than 10 elements.
- The integers are ranged from -2^31 to +2^31 - 1 (they always fit within a 32-bit signed integer).
- 512MB available RAM
- 1 sec timeout

```python
def find_largest(numbers):
    """"
    Takes Python list of integers (1-10 elements inclusive).
    Returns the largest value (int).
    """
    largest = numbers[0]
    for number in numbers:
        if number > largest:
            largest = number
    return largest
```

<i>Remarks: I didn't know the exact number of bits Python delegates to integers, but intuition suggested that 32-bit integer comparisons shouldn't take more than 1s for 10-element array regardless & 512MB should be enough to not requrie any funny business.</i>

## Question 2
<p>You don't remember where you saved the file <u>universe_formula</u>. The only thing you remember about this file is you put it somewhere in a sub-folder of <u>/tmp/documents</u>.</p>
<p>Implement the function <u>locate_universe_formula()</u> to locate <u>universe_formula</u> and then return <b>the absolute path of the file</b> (from the file system root).</p>

- <u>/tmp/documents</u> can contain nested sub-folders and <u>universe_formula</u> can belong to any one of them.
- If <u>universe_formula</u> cannot be found, your program should return <u>None</u>.

```python
import os

def locate_file(directory, file):
    '''
    Conduct depth-first search of some directory.
    Takes directory and file name (str).
    Returns path to file (str).
    '''

    for dirs, files in os.listdir(directory):
        if file in files:
            return os.path(file)
        else:
            for dir in dirs:
                locate_file(dir, file)

    return None

def locate_universe_formula():
    return locate_file('/tmp/documents', 'universe_formula')
```

## Question 3
<p>You work in an automated factory and your objective is to write the function that will dispatch the packages to the correct stack, according to their volume and mass.</p>

- A package is <b>bulky</b> if its volume (Width x Height x Length) is greater than or equal to 1,000,000 cm<sup>3</sup> or when one of its dimension is greater than or equal than 150 cm.
- A package is <b>heavy</b> when its mass is greater or equal to 20 kg.

<p>You must dispatch the packages in the following stacks:</p>

- <u>STANDARD</u>: standard packages (those which are not bulky or heavy) can be handled normally.
- <u>SPECIAL</u>: packages that are either heavy or bulky can't be handled automatically.
- <u>REJECTED</u>: packages that are <b>both</b> heavy and bulky are rejected.

<p>Implement the function <u>solve(width, height, length, mass)</u> (units are centimeter for the dimensions and kilogram for the mass). This function must return a string: the name of the stack where the package should go.</p>

```python
def solve(width, height, length, mass):
    '''
    Classifies packages between stacks.
    Takes three length dimensions width, height, length (int) in cm.
    Takes one mass dimension mass (int) in kg.
    Returns classification of package (str).
    '''
    if width > 150 or height > 150 or length > 150 or width * length * height > 1000000:
        if mass > 20:
            return 'REJECTED'
        else:
            return 'SPECIAL'
    if mass > 20:
        return 'SPECIAL'
    else:
        return 'STANDARD'
```

<i>Remarks: pretty simple if/else conditioning.</i>

## Question 4
<p>Everyday in supermarkets, cashiers have to give back the correct change to customers who pay in cash. How can the determine which bills to give back so that the number of bills is <b>minimal</b>?</p>
<p>In this exercise, we ask you to find the optimal solution to give back change when a checkout only has <b>$2</b>, <b>$5</b>, and <b>$10</b> bills.</p>
<p>To make the problem simpler, we'll consider that we have an <b>infinite amount</b> of bills.</p>
<p>The cash to give back is represented by a dictionary with 3 keys: <b>two</b>, <b>five</b> and <b>ten</b>, which respectively store the number of $2 bills, $5 bills and $10 bills.</p>
<p>Implement the function <b>change(cash)</b> which should return a dictionary whose two, five and ten attributes represent the <b>minimal amount</b> of change to give back.</p>
<p>If it is not possible to give back change, return <b>None</b>.</p>

```python
def change(cash):
    '''
    Takes some change (int).
    Returns the optimal number of $2, $5 and $10 notes (dict).
    Returns None if change is not possible.
    '''
    num_tens = cash // 10
    num_fives = (cash % 10) // 5
    num_twos = ((cash % 10) % 5) // 2

    if ((cash % 10) % 5) % 2 != 0:
        return None
    else:
        return {'two': num_twos, 'five': num_fives, 'ten': num_tens}
```

<i>Remarks: Optimal solution is given by a series of modulus operations.</i>
